```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, classification_report
```

```python
data = load_iris()
```

```python
#2. Load the dataset

daat = load_iris()
#data = load_iris()
x = pd.DataFrame(daat.data, columns=daat.feature_names)
y = pd.Series(daat.target)
x
```

```python
print(f"Shape of data:{x.shape}")
print(f"Classes: {data.target_names}")
print(f"Count of each class: {y.value_counts()}")
```

```python
#Visualize two features
plt.scatter(x.iloc[:,2], x.iloc[:,3], c=y)
plt.xlabel(data.feature_names[2])
plt.ylabel(data.feature_names[3])
```

```python
#3. Split into training and testing sets (80% train, 20% test)
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
#4. Featue scaling
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
#Step 5: Create and train SVM model
model = SVC(kernel ='linear')
model.fit(x_train, y_train)
```

```python
#6. Predict test data
y_pred = model.predict(x_test)
```

```python
#7. Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
#8. Display confusion Matrix
cm = confusion_matrix(y_test, y_pred)
display = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=data.target_names)
display.plot()
plt.show()
```

```python
#9. Display Classification report
print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=data.target_names))
```
